```python
import pandas as pd
from kafka3 import KafkaProducer
from time import sleep
import json
import os
from pymongo import MongoClient


# Kafka setup
hostip = "host.docker.internal"
producer = KafkaProducer(
    bootstrap_servers=[f'{hostip}:9092'],
    api_version=(0, 10)
)

# Load data
df = pd.read_csv("/home/student/work/ass2/data/camera_event_A.csv")

# Kafka topic for Producer A
topic = "camera_event_a"
batch_interval = 5  # seconds

# Send batches by batch_id
for batch_id in sorted(df['batch_id'].unique()):
    batch = df[df['batch_id'] == batch_id]

    for _, row in batch.iterrows():
        # Format timestamp properly
        row['timestamp'] = pd.to_datetime(row['timestamp']).isoformat()

        # Convert to dict and append producer metadata
        msg = row.to_dict()
        msg['producer'] = 'A'
        msg['producer_timestamp'] = pd.Timestamp.now().isoformat()

        # Send message
        producer.send(topic, value=json.dumps(msg).encode('utf-8'))

    producer.flush()
    print(f" Sent batch {batch_id} with {len(batch)} records from Producer A")
    sleep(batch_interval)
```
