### Scenario Background

Malaysia’s road network consistently ranks among the busiest and most accident‑prone in Southeast Asia. Federal roads alone account for a significant proportion of traffic incidents, particularly during peak travel periods and festive seasons, when speed limits of up to 90 km/h (and 110 km/h on expressways) are frequently exceeded in an effort to cover long distances quickly. Since 2012, the Automated Enforcement System (AES) has deployed static speed‑light and red‑light cameras at fixed points to deter speeding and dangerous cornering. However, these point‑capture devices suffer from well‑documented loopholes: drivers can simply decelerate when approaching a camera and then accelerate immediately afterward, rendering enforcement uneven and often ineffective.

To address these shortcomings, the Malaysian Government has begun rolling out the Automated Awareness Safety System (AWAS), a point‑to‑point average‑speed enforcement mechanism (Jamil et al., 2022). Figure 1 illustrates an overview of the AWAS system. AWAS leverages pairs of Ekin Spotter modular cameras equipped with 360° video surveillance and Automatic Number Plate Recognition (ANPR) to record each vehicle’s passage at two distinct checkpoints along a highway segment. By logging the exact timestamps at “Point A” and “Point B,” the system computes the travel time over a known distance (typically 1–5 km) and derives the average speed. Any average exceeding the legal limit (e.g., 110 km/h on expressways) automatically triggers a violation notice, regardless of momentary decelerations.

While AWAS promises more consistent enforcement, it also introduces significant data‑processing challenges. Each camera pair generates a continuous stream of high‑volume events—potentially thousands per minute during peak hours—that must be matched by license plate, ordered by event time, and joined across streams to compute speeds in near real time. The system must tolerate out‑of‑order or late‑arriving events (e.g., network delays), bound state growth via watermarks, and guarantee end‑to‑end exactly‑once processing to prevent duplicate violation records. These requirements make AWAS an ideal case study for a streaming Big Data architecture using Apache Kafka for ingestion, Apache Spark Structured Streaming for stateful stream–stream joins, and MongoDB for scalable storage of both raw events and flagged violations.

Reference:

Jamil, H. M., Shabadin, A., & Ibrahim, M. K. A. (2022). Automated Awareness Safety System (AwAS) for Red Light Running in Malaysia: An Analysis of Four-year Data on Its Effectiveness. Journal of the Society of Automotive Engineers Malaysia, 6(1), 19-29.

#### Description

Referencing the ERD Diagram of Collections from Figure 1.2.1, the Vehicle collection has a "one to many" relationship with the Violation collection. This relationship is derived from the fact that a vehicle can receive distinct violation record/document across each day and each violation record/document is created/flagged based on the violation of speed limits of each vehicle.

The Camera collection has a "many to many" relationship with the Violation collection. This relationship is derived based on the fact that a Vehicle can receive multiple violations in a day from different cameras. However, based on the requirements of the Violation collection, each vehicle can only have one violation record/document in a single day. Therefore, multiple violations from different cameras of a single Vehicle within a single day have to be merged within a single violation record/document which justifies the fact that Violation has a "many to one" relationship with Camera. To justify that Camera has a "many to one" relationship with the Violation, we can derive this from the fact that different vehicles can break the speed limit at different  cameras. Therefore, a violation record/document for different vehicles have to created/updated to include the information of the cameara that detected this violation. In conclusion, this justifies the "many to many" relationship between the Camera and Violation collection.

After the justification of the relationship between collections, this will help us understand the choice of modelling the relationships between the collections using either referencing/embedding.

#### Justification for modelling relationship choice between Vehicle and Violation collection

To model the "one to many" relationship between the Vehicle and Violation collection, we decided to use the referencing approach by referencing the unique identifier "car_plate" attribute from the Vehicle collection in the Violation collection.

##### 1. Read/Write Patterns

For every distinct vehicle that receives a violation in each day, a new document is created in the Violation document. With a N camera setup, there are 2N-1 instances where a vehicle can receive a violation (with a 3 camera setup, there are 5 instances). In respect to the operations performed in the data stream processing and data visualization. There are minimal to no operations that involve reading of the vehicle metadata. Therefore, based on the number of instances that a vehicle can receive a violation from a 3 camera setup and the lack of read operations on the vehicle metadata, we can conclude that the write pattern to the Violation collection is more frequent compared to the read pattern on the Vehicle collection. In conclusion, this justifies that the choice of referencing over embedding is more optimal, since the write pattern is more frequent, this approach avoids rewriting the Vehicle metadata for each collection and since the read pattern is less frequent, we can neglect on the join cost.

##### 2. Data Duplication VS Join Cost

With the choice of referencing Vehicle "car_plate" attribute in the Violation collection, we can significantly reduce duplicated vehicle metadata in the Violation collection since the only attribute that is duplicated within this relationship is the "car_plate" attribute as opposed to the entire Vehicle document. On the flip side, the join cost would be more expensive compared to the embedding approach because the references to the "car_plate" attribute needs to be joined with the Vehicle collection to read the metadata of vehicles. However, as we have established from the Read/Write Patterns above, the join cost can be neglected due to the minimal number of read patterns. Similarly, the nature of the question restricts the number of violation a vehicle can receive in a day and coupled with the fact that a shorter data retention policy can be established, we can significantly reduce the join cost.

##### 3. Consistency

As mentioned in the Read/Write Patterns, the Vehicle collection is a static dataset that is not expected to involve any write operations other than the initial population phase. Therefore, consistency is not something that we have to prioritize because the metadata of vehicles are not expected to change. In a scenario where the metadata of a vehicle is altered, such as the change in the owner's address, referencing will be the optimal choice because the metadata of vehicle's violation need to be accurate and up-to-date. Referencing will allow us to update a single document in the referenced collection (Vehicle) which will immediately reflect all violations received by that vehicle in the Violation collection. In a different scenario where the vehicle is deleted/removed, referencing will not be optimal because all violation documents that involves that specific vehicle needs to be removed as well. However, this is not as detrimental because an appropriate type of join can be used to filter out violations involving the removed vehicle during the processing of the Violation collection.

#### Justification for modelling relationship choice between Camera and Violation collection

To model the "many to many" relationship between the Camera and Violation collection, we used a combination of the embedding and referencing approach instead of a "full reference" approach via the creation of an intermediary collection between the Camera and Violation.

##### 1. Read/Write Patterns

In contrast to the Vehicle Collection, the read pattern on the Camera collection is significantly more frequent due to the join operation in determining whether an event is part of a violation. Based on the frequent read pattern, it may be enticing to choose embedding of the canera document. However, the write pattern of the Violation collection is also as frequent and can be considered as the "costlier" pattern. This is because, the number of Camera document that needs to be written to the Violation collection in the worst case is N * (2 * M - 1) where N is the number of violation document and M is the number of camera document. Therefore, this is a trade-off to our design (further explained later on), where we opt the use of referencing to limit the size of each document in the Violation collection. 

##### 2. Data Duplication VS Join Cost

With the choice of referencing we can significantly reduce the data duplication as shown on the worst-case in the Read/Write Patterns through the referencing attribute "camera_id". In regards to the use of embedding alongside, we know that the relationship between Camera and Violation collection is "Many to Many" and it would make sense to create an intermediary collection like "Event" that makes use of the "camera_id" and "violation_id" attribute to achieve a "full referencing" approach. However, the flaw of this design is that we are introducing additional join-cost to acquire data about the events such as the "timestamp" and the "speed". With the use of embedding we can avoid this join cost with no drawbacks on the data-duplication because each event is unique for each violation. Therefore, the creation of an additional "Event" (intermediary) collection will introduce more complexity. In conclusion, the use of referencing minimizes data duplication between Violation and Camera. However, the join-cost of Violation and Camera can still be expensive but could be improved with efficient join algorithms that can take advantage on the fact that the Camera collection is small. The embedding approach of the Event data in Violation collection removes join-cost between Violation and Event collection without the drawback of data duplication.

##### 3. Consistency

Similar to the Vehicle collection, the Camera collection is a static dataset that is not expected to involve any write operatoins other than the initial population phase. Therefore, consistency is not something that we have to prioritize because the metadata of camera are not expected to change. In a scenario where the metadata of a camera is altered, such as the change in the camera's position, referencing will be the optimal choice because the metadata of camera involved in violations need to be accurate and up-to-date. Referencing will allow us to update a single document in the referenced collection (Camear) which will immediately reflect all violations involving that camera in the Violation collection. In a different scenario where the camera is deleted/removed, referencing will not be optimal because all violation documents that involves that specific camera needs to be removed as well. However, this is not as detrimental because an appropriate type of join can be used to filter out violations involving the removed camera during the processing of the Violation collection.

Adding on to the benefit for using embedding of Event data in the Violation collection, this design of the data model will allow us to create/update/delete informaion of the event data without performing additional operations to ensure that the data is consistent because it is in a denomalized form.

#### Consistency and Idempotency

To determine whether our model supports idempotent writes, be begin by evaluating how each collection supports idempotent writes.

For the Vehicle and Camera Collection, each vehicle is uniquely identified with the "car_plate" attribute and each camera is uniquely identified with the "camera_id" attribute. The purpose of specifying unique indexes as shown on Task 1.1, is to not only illustrate the idea that these attributes are unique identifiers. Similar to the update operation, where mutiple updates to an existing vehicle/camera document would always end up with the same results because there can only be one document that corresponds to that index, so the same document would be updated repeatedly. In conclusion, with the use of unique indexes we can ensure that the Vehicle and Camera collection supports idempotent writes.

For the Violation Collection, we illustrated an upsert pattern in the Violation collection by showing two scenarios of a write operation in Figure 1.3.1 and Figure 1.3.2 that updates an existing violation document and inserts a new violation document. The nature of this upsert pattern is because of the fact that a specific vehicle can only receive one violation document at each day. Therefore, any additional violations that is received by that specific vehicle on that same day would need to be merged with the existing violation document for that specific vehicle on that same day. In simple terms, Figure 1.3.1 shows how the write operation of the document representing the "first violation received by the vehicle on that day" would result in an insert operation on the Violation collection. On the other hand, Figure 1.3.2 shows how the write operation of the document representing the "second violation received by the vehicle on that day" would result in an update operation on the Violation collection.

Now that we have discussed the upsert pattern in the Violation collection, we will be able to see how it affects the idempotency of the Violation collection. Firstly, each violation is uniquely identified with the "car_plate" and "violation_date" attribute (based on the restriction of the AWAS system). This corresponds with Index #1 of the Violation collection and it is a partial design to ensure idempotent writes to the Violation collection. For instance, let's say we have a specific violation document that represents a specific vehicle receiving a specific violation on a specific day. Multiple write operations of the same document is executed to the Violation collection, we know that the Violation collection has an upsert pattern for write operations (explained above). Therefore, Index #1 will enforce a unique constraint to prevent this write operation from inserting duplicate violation documents to the Violation collection. However, due to the upsert pattern of the Violation collection, this write operation can also be an update operation. At first glance, we can simply just update (merge) new violation events with an existing violation document that has the same corresponding "car_plate" and "violation_date" attributes. However, this will cause a non-idempotent write because repeated/non-unique violations will be updated/merged. Therefore, our design of embedding violation events would not achieve a full idempotent write on the Violation collection.
```python
def upsert_violation_collection(violation_collection_instance: pymongo.collection, violation_records: List[Tuple[Dict[str, any], Dict[str, any]]]) -> None:
    '''
    Parameters:
    1) violation_collection_instance
    - Instance of the MongoDB Violation Collection.
    2) violation_records
    - List of tuples in the form of (violation_record, violation_event) to be upserted into the Violation collection.
        
    Description:
    This function handles bulk & idempotent write-operations to the Violation collection.
    Any individual write operations that are failed during the bulk write is extracted and retried.
    
    1) Bulk Upsert:
    
    Purpose:
    Streams of camera event data are processed & stored in batches.
    Therefore, batches of violation data needs to be ingested efficiently.
    
    Implementation:
    Batches of violation data can belong to different violation records. 
    Therefore, a simple updateMany function will not suffice.
    Firstly, records of Violation data is gathered and passed into the UpdateOne constructor.
    The bulk_write function is used to perform a series of write operations as a single transaction/operation.
    
    References: (1), (2)

    2) Idempotent Write:
    
    Purpose: 
    When joining stream data, duplication can occur.
    In our case, violation events can be duplicated across different batches of stream joins.
    Previously we were able to ensure idempotent writes within document level with unique key constraints.
    However, violation events are embedded within documents and unique key constraints only work on the document level.
    
    References: (3)
    
    Implementation:
    The operator $addToSet ensures that embedded objects pushed into the array of the document will be unique.
    
    Rerences: (4)
    
    References:
    (1) https://stackoverflow.com/questions/5292370/fast-or-bulk-upsert-in-pymongo
    (2) https://pymongo.readthedocs.io/en/stable/examples/bulk.html
    (3) https://spark.apache.org/docs/latest/structured-streaming-programming-guide.html#streaming-deduplication
    (4) https://www.mongodb.com/docs/manual/reference/operator/update/addToSet/
    '''
    #initialize upsert operations
    bulk_upserts = [ 
        UpdateOne(
            { 
                "car_plate": violation_record['car_plate'],
                "violation_date": violation_record['violation_date'] 
            },
            {
                "$setOnInsert": violation_record,
                "$addToSet": { "violations" : violation_event },
            },
            upsert = True
        )
        for violation_record, violation_event in violation_records
    ]
    if bulk_upserts:
        try:
            #execute upsert operations in bulk
            violation_collection_instance.bulk_write(bulk_upserts, ordered = False)
        except errors.BulkWriteError as bwe:
            #extract and retry failed write operations
            idx_failed_write_operations = [err['index'] for err in bwe.details.get("writeErrors")]
            if idx_failed_write_operations:
                violation_collection_instance.bulk_write([bulk_upserts[idx] for idx in idx_failed_write_operations], ordered = False)
```

```python
os.environ['PYSPARK_SUBMIT_ARGS'] = (
    '--packages '
    'org.apache.spark:spark-sql-kafka-0-10_2.12:3.3.0,'
    'org.mongodb.spark:mongo-spark-connector_2.12:3.0.1 '
    'pyspark-shell'
)

spark = SparkSession.builder \
    .appName("AWAS Streaming App") \
    .master("local[*]") \
    .getOrCreate()
```

```python
schema = StructType() \
    .add("event_id", StringType()) \
    .add("batch_id", IntegerType()) \
    .add("car_plate", StringType()) \
    .add("camera_id", IntegerType()) \
    .add("timestamp", StringType()) \
    .add("speed_reading", FloatType()) \
    .add("producer_id", StringType())

def read_topic_stream(topic_name: str):
    return spark.readStream.format("kafka") \
        .option("kafka.bootstrap.servers", f"{host_ip}:9092") \
        .option("subscribe", topic_name) \
        .load() \
        .selectExpr("CAST(value AS STRING) as json_str") \
        .select(from_json(col("json_str"), schema).alias("data")) \
        .select("data.*") \
        .withColumn("event_time", to_timestamp(col("timestamp")))

def add_prefix(stream, prefix: str):
    return stream.select([col(column_name).alias(f"{prefix}_{column_name}") for column_name in stream.columns])

camera_stream_a = add_prefix(read_topic_stream("camera_event_a"), "A").withWatermark("A_event_time", "10 minutes")
camera_stream_b = add_prefix(read_topic_stream("camera_event_b"), "B").withWatermark("B_event_time", "10 minutes")
camera_stream_c = add_prefix(read_topic_stream("camera_event_c"), "C").withWatermark("C_event_time", "10 minutes")
```

```python
camera_df = spark.read.format("mongo") \
    .option("uri", f"mongodb://{host_ip}:27017/fit3182_a2_db.Camera") \
    .load()

def get_camera_metadata(camera_id: int):
    return camera_df.filter(camera_df["camera_id"] == camera_id).first().asDict()

camera_a_meta = get_camera_metadata(1)
camera_b_meta = get_camera_metadata(2)
camera_c_meta = get_camera_metadata(3)
```

```python
camera_stream_a_b_window = ceil(((camera_b_meta["position"] - camera_a_meta["position"]) / camera_b_meta["speed_limit"]) * 3600)
camera_stream_b_c_window = ceil(((camera_c_meta["position"] - camera_b_meta["position"]) / camera_c_meta["speed_limit"]) * 3600)
```

```python
camera_stream_a_b = camera_stream_a.join(
    camera_stream_b,
    expr(
        f"""
        A_car_plate == B_car_plate AND
        B_event_time BETWEEN A_event_time AND A_event_time + INTERVAL {camera_stream_a_b_window} SECONDS
        """
    ),
    "inner"
).withColumn("average_speed", (camera_b_meta["position"] - camera_a_meta["position"]) / ((col("B_event_time").cast("double") - col("A_event_time").cast("double")) / 3600))

camera_stream_b_c = camera_stream_b.join(
    camera_stream_c,
    expr(
        f"""
        B_car_plate == C_car_plate AND
        C_event_time BETWEEN B_event_time AND B_event_time + INTERVAL {camera_stream_b_c_window} SECONDS
        """
    ),
    "inner"
).withColumn("average_speed", (camera_c_meta["position"] - camera_b_meta["position"]) / ((col("C_event_time").cast("double") - col("B_event_time").cast("double")) / 3600))
```

```python
def violation_batch_write_mongodb(batch_df, epoch_id, violation_config: dict):
    violations = [row.asDict() for row in batch_df.collect()]
    violation_type = violation_config["violation_type"]
    formatted_violations = []
    if violation_type == "average":
        start_producer_id = violation_config["start_producer_id"]
        end_producer_id = violation_config["end_producer_id"]
        for violation in violations:
            violation_record = {
                "car_plate": violation.get(f"{end_producer_id}_car_plate"),
                "violation_date": violation.get(f"{end_producer_id}_event_time").replace(hour=0, minute=0, second=0, microsecond=0)
            }
            violation_event = {
                "violation_type": violation_type,
                "start_camera_id": violation.get(f"{start_producer_id}_camera_id"),
                "end_camera_id": violation.get(f"{end_producer_id}_camera_id"),
                "start_timestamp": violation.get(f"{start_producer_id}_event_time"),
                "end_timestamp": violation.get(f"{end_producer_id}_event_time"),
                "average_speed": violation.get("average_speed")
            }
            formatted_violations.append((violation_record, violation_event))
        upsert_violation_collection(violation_collection, formatted_violations)
    elif violation_type == "instantaneous":
        producer_id = violation_config["producer_id"]
        for violation in violations:
            violation_record = {
                "car_plate": violation.get(f"{producer_id}_car_plate"),
                "violation_date": violation.get(f"{producer_id}_event_time").replace(hour=0, minute=0, second=0, microsecond=0)
            }
            violation_event = {
                "violation_type": violation_type,
                "camera_id": violation.get(f"{producer_id}_camera_id"),
                "timestamp": violation.get(f"{producer_id}_event_time"),
                "instantaneous_speed": violation.get(f"{producer_id}_speed_reading"),
            }
            formatted_violations.append((violation_record, violation_event))
        upsert_violation_collection(violation_collection, formatted_violations)
    else:
        raise Exception(f"Invalid Violation Type: {violation_type}")
```

```python
from folium.plugins import MarkerCluster
import folium

def generate_map():
    fmap = folium.Map(location=[-37.81, 144.96], zoom_start=14) # need to change this somewhere
    marker_cluster = MarkerCluster().add_to(fmap)

    # Camera markers
    camera_positions = {}
    for cam in camera_coll.find():
        cam_id = cam["camera_id"]
        lat, lon = cam["lat"], cam["lon"]
        camera_positions[cam_id] = (lat, lon)
        folium.Marker([lat, lon], popup=f"Camera {cam_id}",
                      icon=folium.Icon(color="blue", icon="camera")).add_to(marker_cluster)

    # Violation lines
    pipeline = [
        {"$group": {
            "_id": {"start": "$camera_id_start", "end": "$camera_id_end"},
            "count": {"$sum": 1}
        }}
    ]
    edges = list(violation_coll.aggregate(pipeline))
    for edge in edges:
        start_id, end_id = edge["_id"]["start"], edge["_id"]["end"]
        count = edge["count"]
        if start_id in camera_positions and end_id in camera_positions:
            folium.PolyLine(
                [camera_positions[start_id], camera_positions[end_id]],
                color="red" if count > 5 else "orange",
                weight=3,
                tooltip=f"{start_id} → {end_id}: {count} violations"
            ).add_to(fmap)  # hotspot 

    fmap.save("violation_map.html")
```
